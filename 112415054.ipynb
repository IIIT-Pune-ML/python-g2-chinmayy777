## Types and Functions

Extend the function to take an optional 3rd argument.






Extend the function to take another optional flag parameter and add only if it is true






```python
# NOTE: Python has no function overloading. If we write three defs with the same
# name, only the LAST one survives -- the earlier two are silently discarded.
# So the "extended" versions are built with DEFAULT ARGUMENTS instead.

def add(x, y):
    """Original two-argument version."""
    return x + y


def add(x, y, z=0):
    """Extended with an optional 3rd argument (defaults to 0 so add(x, y) still works)."""
    return x + y + z


def add(x, y, z=0, flag=None):
    """Extended again with an optional flag: z is added only when flag is true."""
    if flag:
        return x + y + z
    return x + y


print(add(2, 3))                    # flag not set -> z ignored -> 5
print(add(2, 3, 4))                 # flag still not set -> 5
print(add(2, 3, 4, flag=True))      # flag true -> z added -> 9
print(add(2, 3, 4, flag=False))     # flag false -> z ignored -> 5
```

Assign a function to a variable 'a' and print the output

```python
# A function is just an object, so it can be bound to another name.
a = add                     # 'a' now refers to the function itself (no parentheses!)
print(a)                    # <function add at 0x...>
print(a(2, 3, 4, flag=True))  # calling through the new name

# Assigning the RESULT of a call is a different thing entirely:
b = add(2, 3)
print(b)
```

Show how to use a iterator. First, use enumerate to create a new tuple from above list

```python
# Iterator that starts at 25 and squares the value on every subsequent call.
# The original stub read self.a but never created it -- __init__ was missing.

class MyInit:
    def __init__(self, start=25):
        self.a = start          # initialisation

    def __iter__(self):         # makes the object iterable
        return self

    def __next__(self):         # does the operation
        x = self.a              # value to hand back this time
        self.a = self.a ** 2    # prepare the next value
        return x


obj1 = MyInit()
itr = iter(obj1)

print(next(itr))   # 25
print(next(itr))   # 625
print(next(itr))   # 390625
```

```python
# First, use enumerate to create a new tuple from the list above
listSequence = ["eat", "sleep", "repeat"]

tup = ()
# enumerate() yields (index, item) pairs; we collect them with a for loop
for pair in enumerate(listSequence):
    tup = tup + (pair,)

print(tup)
print(tuple(enumerate(listSequence)))   # the same thing in one step
```

Now use a zip iterator to do the same.

```python
# The same result using zip: pair the list against its own index range
tupZip = tuple(zip(range(len(listSequence)), listSequence))
print(tupZip)

print(tup == tupZip)   # True -- identical output, different mechanism
```

Now state the differences between a zip and enumerate iterator.

Python`enumerate()` and `zip()` are built-in functions that can be used on sequences.

`enumerate()` function can be use to iterate over indices and items of a list.

`zip()` function make an iterator that aggregates elements from each of the iterables. In simple words it can be used to iterate over two or more lists in parallel.









## Strings in Python

Open the file poem.txt and print lenght of each line.

```python
# poem.txt starts with a UTF-8 BOM and uses Windows (CRLF) line endings.
# encoding='utf-8-sig' strips the BOM; rstrip() removes the trailing \r\n so the
# lengths we measure are the real text lengths.

lineLengths = []
with open('poem.txt', 'r', encoding='utf-8-sig') as file:
    for line in file:
        lineLengths.append(len(line.rstrip('\r\n')))

print("Total lines:", len(lineLengths))
print("Length of the first 10 lines:")
for i in range(10):
    print("  line", i + 1, "->", lineLengths[i])
```

Find sum of each line, mean, median and standard deviation of all lines in the file.

```python
import statistics

print("Sum of all line lengths :", sum(lineLengths))
print("Mean line length        :", statistics.mean(lineLengths))
print("Median line length      :", statistics.median(lineLengths))
print("Standard deviation      :", statistics.stdev(lineLengths))
```

Print the last word from the file and change the second last characted in the string to 'g'. 

```python
lastWord = words[-1]
print("Last word:", lastWord)

# Attempt to change the second-last character to 'g'
try:
    lastWord[-2] = 'g'
    print("Changed:", lastWord)
except TypeError as e:
    print("Failed with TypeError:", e)

# The workaround: build a NEW string instead of editing the old one
newWord = lastWord[:-2] + 'g' + lastWord[-1]
print("New string:", newWord)
print("Original is untouched:", lastWord)
```

State if the above operation completed successfully. State reasons

**No, the operation did not complete successfully.**

The assignment `lastWord[-2] = 'g'` raises `TypeError: 'str' object does not support item assignment`.

Strings in Python are **immutable** -- once created, the character data cannot be modified in place. Every "modification" of a string actually produces a brand-new string object, which is why the workaround above uses slicing and concatenation (`lastWord[:-2] + 'g' + lastWord[-1]`) to build a new value while the original stays exactly as it was.

Lists, by contrast, are mutable, which is why `myList[i] = myList[i] ** 2` worked earlier.

Take a list of 10 elements and convert into json using  json library. Demonstrate use of loads and dumps

```python
import json

data = [1, 2, 3, 4, 5, 'six', 'seven', 8.0, True, None]

# dumps -> "dump string": Python object  ->  JSON text
jsonText = json.dumps(data)
print("dumps  :", jsonText)
print("type   :", type(jsonText))

# loads -> "load string": JSON text  ->  Python object
backAgain = json.loads(jsonText)
print("loads  :", backAgain)
print("type   :", type(backAgain))
print("round trip preserved:", backAgain == data)
```

# Working with CSV

Let's import a file with fuel economy data

```python
import csv

with open('mpg.csv') as csvfile:
    mpg = list(csv.DictReader(csvfile))

print("Number of rows:", len(mpg))
print("First row:", mpg[0])
```

Find out how many items are there in the dictionary

```python
for item in mpg[1].items():
    print(item)

print()
print("Number of items in one dictionary:", len(mpg[1]))
```

Find the column names with keys

```python
print(mpg[0].keys())

# The first key is an EMPTY STRING: the csv has an unnamed index column whose
# header cell is blank, so csv.DictReader uses '' as its key.
print()
for col in mpg[0].keys():
    print(repr(col))
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```python
# All values in the dictionaries are strings, so each must be cast to float first.
avgCty = sum(float(d['cty']) for d in mpg) / len(mpg)
print("Average city fuel economy:", avgCty)
```

Find the average hwy fuel economy across all cars

```python
avgHwy = sum(float(d['hwy']) for d in mpg) / len(mpg)
print("Average highway fuel economy:", avgHwy)
```

Use set to return the unique values for the number of cylinders the cars.

```python
# The stub said set(d['cyl']) -- that would iterate the CHARACTERS of a single
# string. We need a generator across every dictionary in mpg, as with 'class' below.
cylinders = set(d['cyl'] for d in mpg)
print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
MpgByCyl = []

for c in cylinders:               # iterate over all the cylinder levels
    summpg = 0
    typecount = 0
    for d in mpg:                 # iterate over all dictionaries
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1
    MpgByCyl.append((c, summpg / typecount))

MpgByCyl.sort(key=lambda x: x[0])   # sort by number of cylinders
MpgByCyl
```

Return the unique values for the class types in our dataset using set.

```python
vehicleclass = set(d['class'] for d in mpg)
print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```python
MileageByClass = []

for t in vehicleclass:            # iterate over all the vehicle classes
    sumpg = 0
    classCount = 0
    for d in mpg:
        if d['class'] == t:
            sumpg += float(d['hwy'])
            classCount += 1
    MileageByClass.append((t, sumpg / classCount))

MileageByClass.sort(key=lambda x: x[1])

MileageByClass
```
